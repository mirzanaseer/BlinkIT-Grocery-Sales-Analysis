```python
df['Item Weight'] = df['Item Weight'].fillna(
    df['Item Weight'].median()
)
```

```python
print("Missing values after treatment:")
print(df['Item Weight'].isnull().sum())
```

```python
plt.figure(figsize=(10, 3))

sns.boxplot(x=df['Item Weight'])

plt.title('Distribution of Item Weight')
plt.xlabel('Item Weight')
plt.tight_layout()
plt.show()
```

```python
df['Item Fat Content']=df['Item Fat Content'].replace({'LF':'Low Fat', 'low Fat':'Low Fat', 'reg':'Regular'})
```

```python
sales_by_category = (
    df.groupby("Item Type")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(sales_by_category)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=sales_by_category.index,
    y=sales_by_category.values
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
outlet_sales = (
    df.groupby("Outlet Type")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(outlet_sales)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=outlet_sales.index,
    y=outlet_sales.values
)

plt.title("Total Sales by Outlet Type")
plt.xlabel("Outlet Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

```python
correlation = df[
    ["Item Visibility", "Item Weight", "Rating", "Sales"]
].corr()

print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".3f"
)

plt.title("Correlation Between Product Characteristics and Sales")

plt.tight_layout()
plt.show()
```

```python
location_sales = (
    df.groupby("Outlet Location Type")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

size_sales = (
    df.groupby("Outlet Size")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Location:")
print(location_sales)

print("\nSales by Outlet Size:")
print(size_sales)
```

```python
product_sales = (
    df.groupby("Item Identifier")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products:")
print(product_sales.head(10))

print("\nBottom 10 Products:")
print(product_sales.tail(10))
```
